```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd
import scipy
import PIL
import requests
```

```python
np.random.seed(613)
x = np.arange(50)
y1 = np.random.randint(0,100,50)
y2 = np.random.randint(0,100,50)
```

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y1)
ax.plot(x,y2)
```

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y1,label='Person1')
ax.plot(x,y2,label='Person2')
ax.legend(loc='lower right')
fig.show()
```

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.plot(x,y1,label='Person1')
ax.plot(x,y2,label='Person2')
ax.legend(loc='lower right', bbox_to_anchor = (1,1))
fig.show()
```

Annotations with the text() function

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label='Person1')
ax.scatter(x,y2,label='Person2')
ax.legend(loc='lower right')

ax.text(20,80, "This value is improtant")
fig.show()
```

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label='Person1')
ax.scatter(x,y2,label='Person2')
ax.legend(loc='lower right')

ax.text(10,80, "This value is improtant!",
ha = 'center',  #align text
color= 'red', #modify font color 
size = 20)   #modify fonsize
fig.show()
```

Positioning Text Annotations

```python
fig, ax = plt.subplots()
ax.axis([0, 10, 0, 10])
ax.text(1, 5 ,
". Data:(1,5)", transform=ax.transData)

ax.text(0.5 , 0.1, ". Axes:(0.5,0.1)", transform=ax.transAxes)

ax.text(0.2 , 0.2, ". Figure:(0.2,0.2)", transform=fig.transFigure)
```

Annotating with Arrows

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label='Person1')
ax.scatter(x,y2,label='Person2')
ax.legend(loc='lower right')

ax.annotate("This value is improtant!", xy = (10,5), xytext=(20,70),
arrowprops = dict(facecolor= 'black'))
fig.show()
```

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label='Person1')
ax.scatter(x,y2,label='Person2')
ax.legend(loc='lower right')

ax.annotate("This value is improtant!", xy = (10,95), xytext=(20,94),
arrowprops = dict(arrowstyle = 'fancy',
                    color= 'hotpink'))
fig.show()
```

Removing tick marks and labels

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label='Person1')
ax.scatter(x,y2,label='Person2')
ax.legend(loc='lower right')

ax.yaxis.set_major_locator(plt.NullLocator()) # removes labels and tick marks
ax.xaxis.set_major_formatter(plt.NullFormatter())  # removes labels only


fig.show()
```

Limiting tick marks

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label='Person1')
ax.scatter(x,y2,label='Person2')
ax.legend(loc='lower right')

ax.xaxis.set_major_locator(plt.MaxNLocator(3))
fig.show()
```

Tick marks intervals

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label='Person1')
ax.scatter(x,y2,label='Person2')
ax.legend(loc='lower right')

ax.xaxis.set_major_locator(plt.MultipleLocator(5))
fig.show()
```

Rotating axis labels

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label='Person1')
ax.scatter(x,y2,label='Person2')
ax.legend(loc='lower right')

plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=50, ha='right')  #ha attaches the right side of the label to the tick mark
fig.show()
```

```python
fig,ax=plt.subplots(figsize=(5,3))
ax.scatter(x,y1,label='Person1')
ax.scatter(x,y2,label='Person2')
ax.legend(loc='lower right')

plt.xticks(rotation=45, ha='right')
plt.xlabel('Shiny New X Axis!', fontsize=18, fontdict={'family':'serif', 'color':'indigo'})
plt.yticks(rotation=50, ha='right')  #ha attaches the right side of the label to the tick mark
fig.show()
```
