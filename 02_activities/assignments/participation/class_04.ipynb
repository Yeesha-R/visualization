```python
fig , (ax1, ax2) = plt.subplots(ncols=2, figsize=(7,3))

ax1.scatter(x1, y1)
ax2.bar(x2, y2)

ax1.set_title("People vs number")
ax2.set_title('Scatter plots of what i don\'t know yet')
ax1.annotate("This value is improtant!", xy = (10,50), xytext=(20,60),arrowprops = dict(facecolor= 'black'))
fig.show()
```

Multiple Visualization on one axes

```python
x = np.array(['Luffy', 'Zoro', 'Nami', 'Usopp', 'Sanji'])
y1 = np.array([110, 180, 240, 99, 220])
y2 = np.array([170, 100, 90, 120, 50])
fig, ax = plt.subplots(figsize=(7,3))

ax.bar (x, y1, color='indigo')
ax.plot(x, y2, color='red')
```

Adding error information to data

```python
y2_sd = np.std(y2)

fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color='red')
```

```python
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color='red')

ax.errorbar(x, y2, 
            yerr= y2_sd, 
            fmt='o', 
            ecolor='indigo', 
            elinewidth=2, 
            capsize=5, 
            capthick=4)
```

To avoid error bars for certain intervals

```python
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color='red')

ax.errorbar(x, y2, 
            yerr= y2_sd, 
            fmt='o', 
            ecolor='indigo', 
            elinewidth=2, 
            capsize=5, 
            capthick=4,
            errorevery=2)
```

Using Images

```python
from PIL import Image
import requests
from io import BytesIO

url = "https://upload.wikimedia.org/wikipedia/en/c/cb/Monkey_D_Luffy.png"

headers = {
    "User-Agent": "Mozilla/5.0",
    "Accept": "image/*,*/*;q=0.8",
    "Referer": "https://en.wikipedia.org/"
}

response=requests.get(url,headers=headers)
image_file=BytesIO(response.content)
image=Image.open(image_file)
image
```

Ading an image to a plot

```python
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color='red')

ax_image = fig.add_axes([0.1, # x coordinate (ON FIGURE, NOT AXES)
                        0.11, # y coordinate (ON FIGURE, NOT AXES)
                        0.15, # image width
                        0.35] # image height
)
```

```python
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color='red')


ax_image = fig.add_axes([0.1, 0.11, 0.15, 0.35])

ax_image.imshow(image)
ax_image.axis('off')
```

Saving Visualizations

```python
fig, ax = plt.subplots(figsize=(7,3))
ax.plot(x, y2, color='red')


ax_image = fig.add_axes([0.1, 0.11, 0.15, 0.35])

ax_image.imshow(image)
ax_image.axis('off')

#path = ('C:/Users/aisra/Downloads/DSI/visualization_001')
#filename='/fig1a.png'
#plt.savefig(path+filename, dpi=300)
```
