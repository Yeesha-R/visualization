```python
import pandas as pd
import matplotlib.pyplot as pyplot
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('ttc-subway-delay-data-since-2025.csv')

df

delay_counts = df['Line'].value_counts()
#print(delay_counts)

df_subway = delay_counts.iloc[0:5]
#print(df_subway)

line_names = {
    'YU': 'Line 1 (Yonge-University)',
    'BD': 'Line 2 (Bloor-Danforth)',
    'SHP': 'Line 4 (Sheppard)',
    'YU/BD': 'Line 1 & Line 2',
    'YUS/BD': 'Line 1 & Line 2'
}

df_subway.rename(index=line_names, inplace=True)
```

```python
plt.figure(figsize=(8,5))
df_subway.plot(kind='bar')
plt.title("Number of Delays by Subway Line")
plt.xlabel('Subway Line')

plt.ylabel('Number of Delays')


plt.tight_layout()
plt.show()
```

```python
ax = df_subway.plot(kind='bar')

ax.set_xticklabels([
    'Line 1\n(Yonge-University)',
    'Line 2\n(Bloor-Danforth)',
    'Line 4\n(Sheppard)',
    'YU/BD',
    'YUS/BD'
], rotation=30)

plt.title("Number of Delays by Subway Line", fontsize=20, fontweight='bold')
plt.xlabel('Subway Line', fontsize=16)
plt.ylabel("Number of Delays", fontsize=16)
plt.tight_layout()
plt.show()
```
